```python
import pandas as pd
import glob

# Cerco in dataraw il file che contiene "migr_acq" nel nome
file = glob.glob("dataraw/*migr_acq*")[0]
print("File trovato:", file)

df = pd.read_csv(file)
print("Righe e colonne:", df.shape)
df.head()
```

```python
# Tengo solo le colonne utili e le rinomino in italiano
df = df[["geo", "TIME_PERIOD", "OBS_VALUE", "OBS_FLAG"]]
df = df.rename(columns={
    "geo": "paese",
    "TIME_PERIOD": "anno",
    "OBS_VALUE": "cittadinanze",
    "OBS_FLAG": "flag"
})
df.head()
```

```python
print("Righe prima:", len(df))

# Tolgo gli aggregati (UE, area euro): sono totali, non paesi
df = df[~df["paese"].str.contains("European Union|Euro area")]

print("Righe dopo:", len(df))
print("Numero di paesi:", df["paese"].nunique())
```

```python
print("Anni: dal", df["anno"].min(), "al", df["anno"].max())
print("Valori mancanti:", df["cittadinanze"].isna().sum())

# Quante righe per ogni flag?
df["flag"].value_counts(dropna=False)
```

```python
# I 5 anni con più cittadinanze acquisite in Italia
df[df["paese"] == "Italy"].sort_values("cittadinanze", ascending=False).head()
```

```python
df.to_csv("dataclean/cittadinanze_albanesi.csv", index=False)
print("File salvato!", len(df), "righe")
```
