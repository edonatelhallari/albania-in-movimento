```python
import pandas as pd
import glob

# Cerco nella cartella dataraw il file che contiene "migr_pop1ctz" nel nome
file = glob.glob("dataraw/*migr_pop1ctz*")[0]
print("File trovato:", file)

# Leggo il file (compresso .gz)
df = pd.read_csv(file)

print("Righe e colonne:", df.shape)
df.head()
```

```python
# Tengo solo le colonne utili e le rinomino in italiano
df = df[["geo", "TIME_PERIOD", "OBS_VALUE", "OBS_FLAG"]]
df = df.rename(columns={
    "geo": "paese",
    "TIME_PERIOD": "anno",
    "OBS_VALUE": "residenti",
    "OBS_FLAG": "flag"
})

print("Paesi:", df["paese"].nunique())
print("Anni: dal", df["anno"].min(), "al", df["anno"].max())
df.head()
```

```python
# Quanti anni di dati ha ogni paese? (il massimo possibile è 28: 1998-2025)
anni_per_paese = df.groupby("paese")["anno"].count().sort_values()
anni_per_paese
```

```python
# Controllo i tipi delle colonne
print(df.dtypes)

# Salvo la tabella pulita
df.to_csv("dataclean/residenti_albanesi_europa.csv", index=False)
print("File salvato!", len(df), "righe")
```
